```python
import csv
#Define the dataset
relation_name="sample_dataset"
attributes=[
    ("age","NUMERIC"),
    ("name","STRING"),
    ("salary","NUMERIC"),
]
instances=[
    (25,"john doe",50000),
    (30,"jane smith",60000),
    (22,"emily davis",45000),
    (29,"michale brown",52000),
]
#Create ARFF file
arff_content=f"@RELATION {relation_name}\n\n"

#Add attributes to ARFF file
for attr_name,attr_type in attributes:
  arff_content+=f"@ATTRIBUTE {attr_name} {attr_type}\n"

#Add data instances to ARFF file
arff_content+="\n@DATA\n"
for instance in instances:
  arff_content+=",".join(map(str,instance))+"\n"

with open("dataset.arff","w") as arff_file:
  arff_file.write(arff_content)

#Create csv file
csv_content=[attr_name for attr_name,_ in attributes]
#write csv file
with open("dataset.csv","w",newline='') as csv_file:
  writer=csv.writer(csv_file)
  writer.writerow(csv_content)
  writer.writerows(instances)
  print("ARFF and CSV files have been created.")
```
